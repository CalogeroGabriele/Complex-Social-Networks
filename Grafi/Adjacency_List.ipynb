```python
# STRUTTURA DEL GRAFO USATO COME ESEMPIO

import matplotlib.pyplot as plt
import networkx as nx

# 1. Creazione del grafo non orientato
G = nx.Graph()
G.add_edges_from([
    ('u', 'v'),
    ('v', 'w'),
    ('u', 'w'),
    ('w', 'z')
])

# 2. Coordinate manuali per replicare esattamente la forma dell'immagine
pos = {
    'v': (0.0, 0.0),
    'u': (0.5, 0.866),  # vertice del triangolo
    'w': (1.0, 0.0),
    'z': (2.0, 0.0)
}

# 3. Etichette degli archi
edge_labels = {
    ('u', 'v'): '$e$',
    ('v', 'w'): '$f$',
    ('u', 'w'): '$g$',
    ('w', 'z'): '$h$'
}

fig, ax = plt.subplots(figsize=(5, 4))

# Disegno dei nodi (bianchi con bordo nero sottile)
nx.draw_networkx_nodes(
    G, pos,
    node_color='white',
    edgecolors='black',
    node_size=1200,
    linewidths=1.5,
    ax=ax
)

# Disegno delle etichette dei vertici in corsivo
labels = {node: f"${node}$" for node in G.nodes()}
nx.draw_networkx_labels(G, pos, labels=labels, font_size=16, font_family='serif', ax=ax)

# Disegno degli archi
nx.draw_networkx_edges(G, pos, edge_color='black', width=1.5, ax=ax)

# Disegno delle etichette degli archi posizionate accanto alla linea
nx.draw_networkx_edge_labels(
    G, pos,
    edge_labels=edge_labels,
    font_size=15,
    font_family='serif',
    rotate=False,
    ax=ax
)

# Regolazione margini e rimozione assi
plt.margins(0.2)
plt.axis('off')

# Salva l'immagine nell'ambiente Colab
plt.savefig('graph_example.png', dpi=300, bbox_inches='tight')
plt.show()
```
