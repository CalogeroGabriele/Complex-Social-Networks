````python
"""
Inserts a new edge within the edgelist.
The function implements two fundamental checks:
1) Makes sure neither combination of the new edge is already inside the data structure.
2) Makes sure the nodes A and B are different from eachother to avoid possibile loops: (u != v).
"""

def insert_edge_edg(graph, u, v):
  if not u or not v or not str(u).strip() or not str(v).strip() or u == v:
    return graph

  if (u,v) not in graph and (v,u) not in graph:
      graph.append((u,v))

  return graph
````

````python
"""
Inserts a new unique vertex at a time within the adjacency list.
"""

def insert_vertex(graph, u):
  if not u or not str(u).strip():
    return graph

  if u not in graph:
    graph[u] = []

  return graph
````

````python
"""
The vertices have been correctly inserted within the data structure,
nonetheless they are still missing the edges highlighting the connections between eachother.
"""

def helper_function_initial_insert_vertexes_adj(adjacency_list):
  insert_vertex(adjacency_list, "Anna")
  insert_vertex(adjacency_list, "Bruno")
  insert_vertex(adjacency_list, "Carla")
  insert_vertex(adjacency_list, "Diego")
  insert_vertex(adjacency_list, "Elena")
  insert_vertex(adjacency_list, "Fabio")
  insert_vertex(adjacency_list, "Giulia")
  insert_vertex(adjacency_list, "Hassan")

  return adjacency_list

helper_function_initial_insert_vertexes_adj(adjacency_list)
````

````python
"""
Checks if an edge between two nodes already exists within the data structure.
Returns True or False depending on the case.
"""

def get_edge_adj(graph, u, v):
  if u in graph and v in graph:
    if v in graph[u] and u in graph[v]:
      return True
    else:
      return False

  return -1
````

````python
"""
Adds a new edge between two nodes within the adjacency list and in addition to the previous check it also looks for possible loops.
"""

def insert_edge_adj(graph, u, v):
  if not u or not v or not str(u).strip() or not str(v).strip() or u == v:
    return graph

  if u not in graph or v not in graph:
    return graph

  if not get_edge_adj(graph, u, v):
    graph[u].append(v)
    graph[v].append(u)

  return graph
````

````python

def helper_function_initial_insert_edges_adj(adjacency_list):
  insert_edge_adj(adjacency_list, "Anna", "Bruno")
  insert_edge_adj(adjacency_list, "Anna", "Carla")
  insert_edge_adj(adjacency_list, "Bruno", "Carla")
  insert_edge_adj(adjacency_list, "Bruno", "Diego")
  insert_edge_adj(adjacency_list, "Carla", "Elena")
  insert_edge_adj(adjacency_list, "Diego", "Elena")
  insert_edge_adj(adjacency_list, "Elena", "Fabio")
  insert_edge_adj(adjacency_list, "Fabio", "Giulia")
  insert_edge_adj(adjacency_list, "Giulia", "Hassan")
  insert_edge_adj(adjacency_list, "Fabio", "Hassan")
  insert_edge_adj(adjacency_list, "Bruno", "Anna")
  insert_edge_adj(adjacency_list, "Anna", "Anna")

  return adjacency_list

helper_function_initial_insert_edges_adj(adjacency_list)
````

````python
"""
Prints all the already existing edges within the structure to obtain a similar view to the one seen
using "insert_edge" on the edgelist.
"""

def edges_adj(graph):
  result = []
  for node1 in graph:
    for node2 in graph[node1]:
      if (node1, node2) not in result and (node2, node1) not in result:
        result.append((node1,node2))

  return result
````

### Edgelist => Task D2 — Validate the edge list


````python
test_edgelist = [ ("Anna","Anna"), ("Anna", "Bruno"), ("Bruno","Anna")   ]

def validate_edge_list(edge_list):
  messages = []
  seen = set()

  for u, v in edge_list:
    #There are no self-loops.
    if u == v:
      messages.append(f"The edge {u,v} is a self-loop")

    #There are no duplicate undirected edges.
    if (u, v) in seen or (v, u) in seen:
      messages.append(f"Duplicate undirected edge found: {(u, v)}")
    else:
      seen.add((u, v))
      seen.add((v, u))


  return messages

validate_edge_list(test_edgelist)
````

### Adjacency List => Task D1 — Validate the adjacency list

````python
test_adjacency_list = {"Anna":["Bruno","Anna","Marco","Marco","Luca"], "Marco":["Anna"], "Luca": [] }

def validate_adjacency_list(adjacency_list):
  messages = []

  vertices = set(adjacency_list.keys())
  seen_vertices = set()


  for key, collaborators in adjacency_list.items():

    #No node is adjacent to itself.
    if key in collaborators:
      messages.append(f"Node {key} appears in its own list of collaborators")

    seen_collaborators = set()
    for collaborator in collaborators:

      #Every node appearing in a neighbour list is also a dictionary key.
      if collaborator not in vertices and collaborator not in seen_vertices:
        messages.append(f"Node {collaborator} appears in {key}'s list but is not a dictionary key")
        seen_vertices.add(collaborator)

      #No neighbour list contains duplicates.
      if collaborator in seen_collaborators:
        messages.append(f"Duplicate collaborator '{collaborator}' found in {key}'s list")
      else:
        seen_collaborators.add(collaborator)

        #Every adjacency is symmetric and Your validator must report missing keys without crashing when checking symmetry.
        if collaborator in vertices and collaborator != key and key not in adjacency_list[collaborator]:
          messages.append(f"The adjacency between {key} and {collaborator} is not symmetric")

  return messages

validate_adjacency_list(test_adjacency_list)
````

````python
def sort_input_edges(edges):

  sorted_edges = []

  for u, v in edges:
    sorted_pair = sorted((u,v))
    sorted_edge = tuple(sorted_pair)
    sorted_edges.append(sorted_edge)

  return sorted(sorted_edges)

def compare_representations(edgelist, adjacency_list):
  message = []

  edgelist_sorted_nodes = sorted(vertices_edg(edgelist))
  adjacency_list_sorted_nodes = sorted(vertices_adj(adjacency_list))

  print("Nodes:\n")
  print(f"Nodes edgelist: {edgelist_sorted_nodes}\n")
  print(f"Nodes adj: {adjacency_list_sorted_nodes}")

  if edgelist_sorted_nodes == adjacency_list_sorted_nodes:
    message.append("The edgelist and adjacency list contain the same nodes")
  else:
    message.append("The edgelist and adjacency list do not contain the same nodes.")

  print("\n------------------------------------\n")

  edgelist_sorted_edges = sort_input_edges(edgelist)
  adjacency_list_sorted_edges = sort_input_edges(edges_adj(adjacency_list))

  if edgelist_sorted_edges == adjacency_list_sorted_edges:
    message.append("The edgelist and adjacency list contain the same edges")
  else:
    message.append("The edgelist and adjacency list do not contain the same edges.")

  print("Edges:\n")
  print(f"Edges edgelist: {edgelist_sorted_edges}\n")
  print(f"Edges adj: {adjacency_list_sorted_edges}\n")

  print("\n------------------------------------\n")

  return message

compare_representations(edgelist, adjacency_list)
````

# Questions and Answers
1) In a Markdown cell, explain this equality $$\sum_{v \in V} \deg(v) = 2|E|$$ by considering how one edge contributes to the degrees of its endpoints.

The idea is that in an undirected graph, every edge connects two nodes. As a result, each single edge contributes +1 to the degree of both nodes it is connected to. Consequently, summing the degrees across all vertices counts every edge exactly twice, which makes the total sum of degrees equal to twice the number of edges.

2) Is this equality alone sufficient to establish that an adjacency list is valid? Explain your reasoning.

The equality is not sufficient to prove that an adjacency list is valid as it is incapable of detecting structural errors or potential graph rules violations.

For example:
* It fails to identify self-loops, which violate the graph rules while artificially increasing degrees and edge counts.
* It cannot detect duplicate collaborators. If a collaborator appears multiple times, that node's degree is artificially inflated; this could be compensated for by missing connections elsewhere in the graph, keeping the overall equality satisfied.
* A listed collaborator might not exist as a primary key in the dictionary, which the formula fails to flag.
* It fails to enforce symmetry between the nodes.
